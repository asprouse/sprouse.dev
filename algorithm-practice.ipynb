```python
def is_palindrome(s: str, i: int, j: int) -> bool:
  while i != j:
    if str[i] != str[j]:
      return False
    i += 1
    j -= 1


  return True



class Solution:
  def longestPalindrome(self, s: str) -> str:
    memo: dict[tuple[int, int], bool] = {}
    max_length = 0
    range: tuple[int, int] | None = None
    n = len(s)
    for i in range(n):
      if i + 1 < n:
        for j in range(i, n):
          if is_palindrome(s, i, j):
            length = j - i;
            if max_length < length:
              range = (i, j)
              max_length = length

    return s[i:j]
```
